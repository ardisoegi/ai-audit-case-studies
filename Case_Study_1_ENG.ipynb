# 2. Analysis

**2.1 - Audit Checklist**
1. Which transactions have an amount above the overall average?
2. How many rows contain missing values (NaN), and in which columns?
3. Which employee appears more than once and is their total transaction amount reasonable?

**2.2 - Checklist 1 Solution**

Which transactions have an amount above the overall average?

**Checklist 1 - Step 1** : Finding the average transaction value

```python
# Method to calculate the mean transaction value
df_mean_transaction = df['amount'].mean()

# Output
print(f"Average transaction value : {df_mean_transaction}")
```

**Checklist 1 - Step 2** : Finding transactions where 'amount' exceeds the average

```python
# Method to filter amount values greater than the mean
df_above_average = df[df['amount'] > df_mean_transaction]

# Output
print(df_above_average)
```

**Checklist 1 - Conclusion**

Found **2 transactions** exceeding the average value:

1. index 2; &nbsp;&nbsp;&nbsp; E003 &nbsp; Andi Wijaya &nbsp; Finance &nbsp; 15,000,000.0 &nbsp; 2024-01-07 &nbsp; approved
2. index 5; &nbsp;&nbsp;&nbsp; E003 &nbsp; Andi Wijaya &nbsp; Finance &nbsp;&nbsp; 9,800,000.0 &nbsp; 2024-01-10 &nbsp; approved

---

**2.3 - Checklist 2 Solution**

How many rows contain missing values (NaN), and in which columns?

**Checklist 2 - Step 1** : Finding missing values in the dataset

```python
# Method to convert empty strings to NaN
df['department'] = df['department'].replace("", np.nan)

# Method to find missing values
df_missing_values = df.isnull().sum()

# Output
print(df_missing_values)
```

**Checklist 2 - Conclusion**

Missing values found: **1** in **amount** column, **1** in **department** column, and **1** in **status** column

---

**2.4 - Checklist 3 Solution**

Which employee appears more than once and is their total transaction amount reasonable?

**Checklist 3 - Step 1** : Finding employees who appear more than once

```python
# Method to group data by namployee_id and count occurrences
df_employee_frequency = df.groupby('employee_id')['name'].value_counts()

# Output
print(df_employee_frequency)
```

**Checklist 3 - Step 2** : Finding employees with unreasonable transaction amounts

```python
# Method to group by employee_id and sum the total amount
df_total_per_employee = df.groupby('employee_id').agg({'name': 'first', 'amount': 'sum'})

# Output
print(df_total_per_employee)
```

**Checklist 3 - Conclusion**

Found employee **E003** name **Andi Wijaya** appearing **2 times** in the transaction log
