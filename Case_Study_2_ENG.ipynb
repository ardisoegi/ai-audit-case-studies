# 2. Analysis



## **2.1 - Audit Checklist**

1. Do students with scores outside the valid range get an appropriate graduation status?
2. Does the system have input validation to filter non-numeric values or values outside the 0 to 100 scoring scale?
3. Will inconsistent or non-standardized student status labels affect the system's decision output?
4. Is there any bias indication in the scoring system results toward student groups based on gender? Are the average scores and graduation rates distributed proportionally and fairly?

## **2.2 - Checklist 1 Solution**





**Checklist 1 - Step 1** : Finding values outside the 0-100 scale or non-valid values





```python
# Filter students by 'ai_score' to find values outside the 0-100 scale or non-valid
df_invalid_scores = df[
    (df['ai_score'] > 100) |
    (df['ai_score'] < 0) |
    (df['ai_score'].isnull())
]
display(df_invalid_scores)
```

### **Checklist 1 - Conclusion**

Do students with scores outside the valid range get an appropriate graduation status?
>No, because **3 students** were found with scores outside the 0-100 scale or invalid values:
- **S003** with value **NaN**, the system has no value to make a decision, but still produces status **pass**
- **S008** with value **110** exceeding the maximum valid limit, the system still processes and produces **pass**
- **S014** with value **-10** below the minimum valid limit, the system still processes and produces **fail**

## **2.3 - Checklist 2 Solution**

**Checklist 2 - Step 1** : Counting students with scores outside the 0-100 scale or invalid values

```python
# Count students with invalid values
invalid_count = df_invalid_scores.shape[0]
print(f"Number of students with invalid values: {invalid_count}")
```

**Checklist 2 - Step 2** : Counting students with valid scores within the 0-100 scale

```python
valid_count = df.shape[0] - invalid_count
print(f"Number of students with valid values: {valid_count}")
```

### **Checklist 2 - Conclusion**

Does the system have input validation to filter non-numeric values or values outside the 0 to 100 scoring scale?
>No, because **3 students** with scores outside the 0-100 scale or invalid values still received a **graduation status** (**pass** or **fail**). Without input validation, the system cannot distinguish data that is fit to process from data that is not, which means all decisions the system produces for invalid data cannot be trusted.


## **2.4 - Checklist 3 Solution**

**Checklist 3 - Step 1** : Testing label format in the 'status' column




```python
# Count occurrences of each label in 'status' column
display(df['status'].value_counts())
```

**Checklist 3 - Step 2** : Counting occurrences of standard labels (assumed 'pass' and 'fail')

```python
# Filter 'pass' labels that match the standard and count occurrences
df_pass_standard= df[df['status'] =='pass']
display(df_pass_standard)
print(f"Number of rows with standard 'pass' label: {df_pass_standard.shape[0]}\n")
# Filter 'fail' labels that match the standard and count occurrences
df_fail_standard= df[df['status'] =='fail']
display(df_fail_standard)
print(f"Number of rows with standard 'fail' label: {df_fail_standard.shape[0]}\n")
```

### **Checklist 3 - Conclusion**

Will inconsistent or non-standardized student status labels affect the system's decision output?
>Yes, because the **status** column was found to have **4 categories**: **PASS, Pass, pass**, and **fail**. I assume the standard categories are **pass** and **fail** as the graduation benchmark:
- **10 students** with status **pass** passed
- **6 students** with status **fail** did not pass
- **2 students** have non-standard status labels (**1 student** with status **PASS** and **1 student** with status **Pass**)

Based on the output of <code><b>df_pass_standard = df[df['status'] == 'pass']</b></code> in step 2, students with status **PASS** and **Pass** are treated as not passing, because they do not match the system's standard.

So **2 students** who should have passed will be ignored by the system, because their status labels do not match the standard.




## **2.5 - Checklist 4 Solution**

**Checklist 4 - Step 1** : Normalizing non-standard formats for inclusion in the analysis
> *This step is done only for the purpose of the following analysis, not as a fix to the system. Label inconsistency remains a valid finding in checklist 3.*

```python
# Create new column status_normalized to standardize graduation labels
df['status_normalized'] = df['status'].str.lower()
display(df)

```

**Checklist 4 - Step 2** : Finding average 'ai_score' by 'gender'




```python
# Find average 'ai_score' grouped by 'gender' column
display(df.groupby('gender')['ai_score'].agg(count_students='size', avg_ai_score='mean').reset_index())
```

**Checklist 4 - Step 3** : Finding graduation rate by 'gender' percentage.



```python
# Find occurrence count of graduation labels 'status_normalized' grouped by 'gender' column
df_status_percentage = df.groupby('gender')['status_normalized'].value_counts().reset_index()

# Create 'percentage' column from occurrence count
df_status_percentage['percentage'] = (
    df_status_percentage['count'] /
    df_status_percentage.groupby('gender')['count'].transform('sum') * 100
)

display(df_status_percentage)

```

### **Checklist 4 - Conclusion**

Is there any indication of bias in the system's scoring results against student groups based on gender? Are average scores and graduation rates distributed proportionally and fairly?
> There is an indication of gender bias affecting graduation results that needs further investigation;
- The number of **female** students is **8** compared to **male** students at **10**, there is a proportional imbalance in the dataset that affects fairness.
- The average score of **female** students is lower than **male** students, at **70.12** compared to **83.33**. This difference also affects the system's assessment.
- **Female** students have a graduation rate of **pass 50%** vs **fail 50%**, while **male** students have **pass 80%** vs **fail 20%**. Further investigation is needed on the graduation ratio difference by gender to determine if this is systemic bias.

# 3. Audit Report



## **3.1 - Findings**

1. **3 rows** with values **NaN, -10,** and **110** still received a graduation status, even though those values are outside the 0-100 scoring scale. The system risks producing graduation decisions for invalid inputs because there is no input validation. Risk : **High**
2. The **status** column was found to have **4 categories**: **PASS, Pass, pass**, and **fail**. Assuming the standard categories are **pass** and **fail**, students with non-standard categories will be ignored by the system. Risk : **High**
3. The dataset shows high score variation (standard deviation **std: 24**). Investigation by gender shows an uneven gap;
>- The average score of **female** students at **70.12** is lower than **male** students at **83.33**.
>- A gap was also found in graduation rates, **female** students at **pass 50%** compared to **male** students at **pass 80%**.

>- The differences above need further investigation to determine if this reflects actual student ability or indicates systemic bias. Risk : **Medium**




## **3.2 - Recommendations**

1. Input validation needs to be established as a standard requirement so that values in the 'ai_score' column are always valid numbers within the 0-100 scale. This is directed to Developer/Engineer to follow up.
2. Standardization of categories in the status column needs to be applied so that no data is at risk of being ignored by the system due to non-compliance, even when students meet the graduation criteria. This needs to be handled by Developer/Engineer.
3. The data sample for average comparison per gender needs to be balanced in terms of gender representation. The graduation rate per gender is also higher for male students than female students. Further investigation is needed to determine whether the gender distribution difference in the dataset reflects bias in data collection or systemic bias in the model. Directed to the Data Engineer/Data Scientist/AI Ethics Analyst to evaluate the bias in the model.



## **3.3 - Limitations**

1. This dataset contains only 18 rows and was generated with the help of generative AI for audit practice purposes, findings need to be verified with a larger dataset before any conclusion can be drawn.
2. The fairness analysis in checklist 4 uses normalized labels, but the actual system ignores non-standard labels. Graduation rate percentages per gender may differ if calculated based on the actual system behavior.
3. Fairness audit was only conducted based on gender category. The class category was not analyzed because the sample size per class is too small to produce statistically significant conclusions.
4. Average ai_score per gender was calculated with NaN values ignored by default in Pandas, so the number of students counted per gender may not be equal.
5. This dataset does not include documentation on how ai_score was generated, whether it is purely model output, a transformation of original scores, or a combination of both. Audit conclusions need to be verified with complete system documentation.
