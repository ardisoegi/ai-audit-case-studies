1. Apakah siswa yang memperoleh nilai diluar ketentuan mendapatkan status kelulusan yang sesuai?

## **2.2 - Solusi Checklist 1**





**Checklist 1 - Step 1** : Mencari nilai yang diluar skala 0-100 atau nilai non valid





```python
# Mem-filter siswa berdasar kolom 'ai_score' untuk mencari nilai yang tidak sesuai skala 0 - 100 atau non valid
df_nilai = df[
    (df['ai_score'] > 100) |
    (df['ai_score'] < 0) |
    (df['ai_score'].isnull())
]
display(df_nilai)
```

### **Checklist 1 - Conclusion**

Apakah siswa yang memperoleh nilai diluar ketentuan mendapatkan status kelulusan yang sesuai?
>Tidak, karena ditemukan **3 siswa** yang mempunyai nilai diluar skala 0-100 atau nilai tidak valid :
- **S003** dengan nilai **NaN**, sistem tidak memiliki nilai untuk membuat keputusan, tapi tetap menghasilkan status **pass**
- **S008** dengan nilai **110** melebihi batas maksimal valid, sistem tetap memproses dan menghasilkan **pass**
- **S014** dengan nilai **-10** di bawah batas minimum valid, sistem tetap memproses dan menghasilkan **fail**

## **2.3 - Solusi Checklist 2**

**Checklist 2 - Step 1** : Mencari banyaknya siswa yang mendapatkan nilai diluar skala 0-100 atau nilai tidak valid

```python
# Menghitung siswa yang mempunyai nilai tidak valid
nilai_tidak_valid = df_nilai.shape[0]
print(f"Banyaknya siswa yang mempunyai nilai tidak valid : {nilai_tidak_valid}")
```

**Checklist 2 - Step 2** : Mencari banyaknya siswa yang mendapatkan nilai sesuai ketentuan skala 0-100

```python
nilai_valid = df.shape[0] - nilai_tidak_valid
print(f"Banyaknya siswa yang mempunyai nilai valid : {nilai_valid}")
```

### **Checklist 2 - Conclusion**

Apakah sistem memiliki validasi input untuk filtrasi data nilai yang bukan numerik atau nilai yang tidak sesuai ketentuan penilaian skala 0 sampai 100?
>Tidak, karena ditemukan sebanyak **3 siswa** yang mempunyai nilai diluar skala 0-100 atau nilai tidak valid tetap mendapatkan **status kelulusan** (**pass** or **fail**). Jika validasi input tidak ada, sistem tidak bisa membedakan data yang layak diproses atau yang tidak layak, ini berarti semua keputusan yang dihasilkan sistem untuk data tidak valid tidak bisa dipercaya.


## **2.4 - Solusi Checklist 3**

**Checklist 3 - Step 1** : Menguji format label di kolom 'status'




```python
# Menghitung banyaknya kemunculan data label di kolom 'status'
display(df['status'].value_counts())
```

**Checklist 3 - Step 2** : Menghitung banyaknya kemunculan label yang sesuai standart (diasumsikan 'pass' dan 'fail')

```python
# Memfilter label 'pass' yang sesuai standart ketentuan dan jumlah kemunculannya
df_pass_standart= df[df['status'] =='pass']
display(df_pass_standart)
print(f"Banyaknya data berlabel 'pass' yang sesuai standart : {df_pass_standart.shape[0]}\n")
# Memfilter label 'fail' yang sesuai standart ketentuan dan jumlah kemunculannya
df_fail_standart= df[df['status'] =='fail']
display(df_fail_standart)
print(f"Banyaknya data berlabel 'fail' yang sesuai standart : {df_fail_standart.shape[0]}\n")
```

### **Checklist 3 - Conclusion**

Apakah pelabelan status siswa yang tidak konsisten atau tidak diseragamkan akan mempengaruhi hasil keputusan sistem?
>Ya, karena pada kolom **status** ditemukan sebanyak **4 kategori** yaitu; **PASS, Pass, pass**, dan **fail**. Saya asumsikan jika kategori standar yang ditentukan adalah **pass** dan **fail** sebagai patokan kelulusan :
- Sebanyak **10 siswa** dengan status **pass**  diluluskan
- Sebanyak **6 siswa** dengan status **fail** tidak lulus
- **2 siswa** status kelulusannya tidak sesuai standart (**1 siswa** status **PASS** dan **1 siswa** status **Pass**)

Berdasarkan hasil <code><b>df_pass_standart = df[df['status'] == 'pass']</b></code> di step 2, outputnya menganggap siswa berstatus **PASS** dan **Pass** menjadi tidak lulus, karena tidak sesuai ketentuan sistem.

Jadi **2 siswa** yang seharusnya lulus akan diabaikan sistem, disebabkan oleh label kelulusannya tidak sesuai ketentuan.




## **2.5 - Solusi Checklist 4**

**Checklist 4 - Step 1** : Normalisasi format non standart agar dapat ikut dianalisis
> *Langkah ini dilakukan hanya untuk keperluan analisis berikut, bukan perbaikan terhadap sistem. Inkonsistensi label tetap merupakan temuan valid di checklist 3.*

```python
# Membuat kolom baru status_normalized untuk menyeragamkan perlabelan kelulusan
df['status_normalized'] = df['status'].str.lower()
display(df)

```

**Checklist 4 - Step 2** : Mencari rata-rata 'ai_score' berdasarkan 'gender'




```python
# Mencari rata-rata 'ai_score' berdasarkan dari kolom 'gender'
display(df.groupby('gender')['ai_score'].agg(jumlah='size', rata_ai_score='mean').reset_index())
```

**Checklist 4 - Step 3** : Mencari tingkat kelulusan berdasarkan persentase 'gender'.



```python
# Mencari jumlah kemunculan label kelulusan 'status_normalized' berdasarkan dari kolom 'gender'
df_status_percent = df.groupby('gender')['status_normalized'].value_counts().reset_index()

# Membuat kolom 'percentage' dari jumlah kemunculan 'count'
df_status_percent['percentage'] = (
    df_status_percent['count'] /
    df_status_percent.groupby('gender')['count'].transform('sum') * 100
)

display(df_status_percent)

```

### **Checklist 4 - Conclusion**

Apakah ada indikasi bias dari hasil penilaian sistem terhadap kelompok siswa berdasarkan background gender siswa? apakah rata-rata skor dan tingkat kelulusan terdistribusi secara proporsional dan adil?
> Ada indikasi bias terhadap gender siswa yang mempengaruhi kelulusan yang perlu investigasi lebih lanjut;
- Jumlah siswa gender **female** (perempuan) sebanyak **8** dibanding **male** (laki-laki) yaitu **10**, terdapat ketidakseimbangan proporsional dari dataset yang mempengaruhi aspek fairness.
- Rata-rata siswa **female** (perempuan) lebih rendah daripada siswa **male** (laki-laki), yaitu **perempuan 70.12** dibanding **laki-laki 83.33**. Perbandingan rata-rata ini juga mempengaruhi penilaian oleh sistem.
- Siswa **female** (perempuan) persentase kelulusan adalah **pass 50%** dibanding **fail 50%**, sedangkan siswa **male** (laki-laki) memiliki persentase yaitu **pass 80%** dibanding **fail 20%**. Perlu investigasi lanjutan mengenai perbedaan ratio kelulusan berdasar gender agar tidak bias sistemik.
