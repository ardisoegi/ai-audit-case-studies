# 1. Overview


##**1.1 - Konteks Dataset**

Nama Dataset : **log transaksi karyawan**

Dataset ini merupakan data simulasi untuk keperluan latihan audit yang berasal dari bantuan AI generatif. Audit  ini dilakukan untuk mencari mencari anomali yang disengaja dari Dataset

> *Preparation : Untuk memulai, jalankan kode di bawah ini (shortcut : ctrl + Enter)*













```python
import pandas as pd
import numpy as np

# Berikut adalah data log transaksi karyawan
data = {
    "employee_id": ["E001", "E002", "E003", "E004", "E005",
                    "E003", "E006", "E007", "E008", "E009"],
    "name": ["Budi Santoso", "Siti Rahma", "Andi Wijaya", "Dewi Kusuma",
              "Reza Pratama", "Andi Wijaya", "Maya Indah",
              "Fajar Nugroho", "Linda Sari", "Hendra Gunawan"],
    "department": ["Finance", "HR", "Finance", "IT", "Finance",
                   "Finance", "Marketing", "", "HR", "IT"],
    "amount": [1500000, 750000, 15000000, None, 1200000,
               9800000, 500000, 650000, -200000, 900000],
    "date": ["2024-01-05", "2024-01-06", "2024-01-07", "2024-01-08",
             "2024-01-09", "2024-01-10", "2024-01-11", "2024-01-12",
             "2024-01-13", "2024-01-14"],
    "status": ["approved", "approved", "approved", "pending",
                "approved", "approved", "approved", None,
                "approved", "rejected"]
}

df = pd.DataFrame(data)

# Output
print(df)
```

# 2. Analisis




##**2.1 - Checklist Audit**
1.   Transaksi mana yang jumlahnya di atas rata-rata keseluruhan?
2.   Ada berapa baris yang memiliki nilai kosong (NaN) di kolom mana saja?
3.   Employee mana yang muncul lebih dari satu kali dan apakah total transaksinya wajar?

##**2.2 - Solusi Checklist 1**

Transaksi mana yang jumlahnya di atas rata-rata keseluruhan?



**Checklist 1 - Step 1** : Mencari nilai rata-rata

```python
# Metode untuk mencari rata-rata (mean)
df_mean_transaksi = df['amount'].mean()

# Output
print(f"Nilai transaksi rata-rata : {df_mean_transaksi}")
```

**Checklist 1 - Step 2** : Mencari data yang nilai 'amount' melebihi nilai rata-rata


```python
# Metode mencari nilai amount > dari mean
df_atas_rata = df[df['amount'] > df_mean_transaksi]

# Output
print(df_atas_rata)
```


###**Checklist 1 - Conclusion**

Ditemukan **2 transaksi** yang melebihi nilai rata-rata pada :

1. index 2;        E003  Andi Wijaya    Finance  15000000.0  2024-01-07  approved
2. index 5;        E003  Andi Wijaya    Finance   9800000.0  2024-01-10  approved


---




##**2.3 - Solusi Checklist 2**

Ada berapa baris yang memiliki nilai kosong (NaN) di kolom mana saja?


**Checklist 2 - Step 1** : Mencari nilai kosong dataset

```python
# Metode mengubah semua string kosong menjadi NaN
df['department'] = df['department'].replace("", np.nan)

# Metode mencari nilai kosong
df_nilai_kosong = df.isnull().sum()

# Output
print(df_nilai_kosong)
```

###**Checklist 2 - Conclusion**

Jadi, ditemukan nilai kosong di kolom **amount** sebanyak **1**, kolom **department** sebanyak **1**, dan  kolom **status** sebanyak **1**


---



##**2.4 - Solusi Checklist 3**

Employee mana yang muncul lebih dari satu kali — dan apakah total transaksinya wajar?


**Checklist 3 - Step 1** : Mencari nama yang sering muncul

```python
# Metode megelompokkan data nama berdasarkan employee_id
df_karyawan_sering_muncul = df.groupby('employee_id')['name'].value_counts()

# Output
print(df_karyawan_sering_muncul)
```

**Checklist 3 - Step 2** : Mencari nama karyawan berdasarkan ID yang transaksinya tidak wajar, lalu menilai anomali transaksinya

```python
# Metode mengelompokkaan per employee_id dan menjumlah nilai amount
df_total_per_karyawan = df.groupby('employee_id').agg({'name': 'first', 'amount': 'sum'})

# Output
print(df_total_per_karyawan)
```

###**Checklist 3 - Conclusion**

Ditemukan ID **E003** nama **Andi Wijaya** melakukan **2 kali** transaksi
